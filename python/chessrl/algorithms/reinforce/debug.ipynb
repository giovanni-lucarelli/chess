```python
# system
import sys 
sys.path.insert(0, '../../../../')

# utils
import numpy as np 
import logging
from chessrl.utils.load_config import load_config
from chessrl.utils.endgame_loader import get_all_endgames_from_dtz
from chessrl import Env, SyzygyDefender
import numpy as np
import torch
import random
import matplotlib.pyplot as plt

import os
config_path = os.path.join('./', 'config.json')
config = load_config(config_path)
logging.basicConfig(level=config['log_level'], format = '%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

from chessrl.algorithms.policy_gradient.reinforce import Reinforce, get_legal_move_indices, idx_to_move, move_to_idx

def set_seed(s=0):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if torch.cuda.is_available(): torch.cuda.manual_seed_all(s)
set_seed(0)
device = torch.device('cpu')
```

```python
endgames = get_all_endgames_from_dtz(csv_path='../../../../tablebase/krk/krk_train.csv', dtz=3)
train_endgames = [endgames[i]['fen'] for i in range(len(endgames)) if i < 300]

endgames = get_all_endgames_from_dtz(csv_path='../../../../tablebase/krk/krk_test.csv', dtz=3)
test_endgames = [endgames[i]['fen'] for i in range(len(endgames)) if i < 100]
```

```python
agent = Reinforce(
    tb_path='../../../../tablebase/krk/',
    gamma = 1.0,
    lr=0.001,
    hidden=(128,128,64)
    )
```

```python
losses, rewards = agent.train(train_endgames, epochs=30)
agent.save('output/reinforce.pt')
```

```python
def evaluate(policy_path, fens, max_steps=50):
    agent = Reinforce(tb_path="../../../../tablebase/krk/")
    agent.load(policy_path, device=torch.device('cpu'))
    agent.policy.eval()
    wins = 0
    for fen in fens:
        env = Env.from_fen(fen, two_ply_cost=0.0, draw_penalty=1.0, checkmate_reward=1.0,
                           defender=SyzygyDefender(tb_path="../../../../tablebase/krk/"))
        steps = 0
        done = False
        while not done and steps < max_steps:
            legal = get_legal_move_indices(env)
            a_idx = agent.policy.get_action_greedy(env, legal)  # greedy-ish
            uci = idx_to_move[a_idx]
            step = env.step(uci)
            done = step.done
            steps += 1
        if env.state().is_checkmate():
            wins += 1
    return wins / len(fens)
```
