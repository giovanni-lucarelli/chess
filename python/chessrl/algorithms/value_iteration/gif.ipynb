```python
#!/usr/bin/env python3

# system
import sys 
sys.path.insert(0, '../../')

# utils
import os
import glob
import logging 
from chessrl.utils.load_config import load_config
config_path = os.path.join('./', 'config.json')
config = load_config(config_path)
logging.basicConfig(level=config['log_level'], format = '%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)
import matplotlib.pyplot as plt # type: ignore

# chess
from chessrl.env import Env, SyzygyDefender, LichessDefender
from chessrl.utils.io import load_policy_jsonl
from chessrl.utils.plot_chess import plot_game
```

```python
policy = load_policy_jsonl("../../../../artifacts/policies/vi_krk_greedy.jsonl")
fen = "8/3k4/8/3K4/8/8/3R4/8 w - - 0 1"
defender = LichessDefender()
env = Env.from_fen(fen, gamma=0.99, defender=defender)
while True:
    env._apply(policy[env.to_fen()])
    env.ply += 1
    plot_game(env.state(), f'ply{env.ply}.png')
    if env.game.is_game_over():
        break
    new_fen = env.to_fen()
    black_reply = defender.best_reply_uci(new_fen)
    env._apply(black_reply)
    env.ply += 1
    plot_game(env.state(), f'ply{env.ply}_defender.png')
```
